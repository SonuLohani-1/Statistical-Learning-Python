```python
mu_cap = boston['medv'].mean()
mu_cap
```

# (b)

```python
std_err = boston['medv'].std()/np.sqrt(boston.shape[0])
std_err
```

# (c)

```python
def boot_fn(data, idx):
    return np.mean(data.iloc[idx]['medv'])
```

```python
rng = np.random.default_rng(1)
B = 1000 # no. of bootstrap samples
bootstrap_mean = []
for i in range(B):
    idx = rng.choice(boston.shape[0], boston.shape[0], replace=True)
    bootstrap_mean.append(boot_fn(boston, idx))
print(np.mean(bootstrap_mean))
print(np.std(bootstrap_mean))
```

# (d)

```python
(mu_cap-2*std_err, mu_cap+2*std_err)
```

```python
mu_bootstrap = np.mean(bootstrap_mean)
sde_bootstrap = np.std(bootstrap_mean)
(mu_bootstrap - 2*sde_bootstrap, mu_bootstrap + 2*sde_bootstrap)
```

# (f)

```python
def boot_med_fn(data, idx):
    return np.median(data.iloc[idx]['medv'])
```

```python
rng = np.random.default_rng(1)
B = 1000 # no. of bootstrap samples
bootstrap_med = []
for i in range(B):
    idx = rng.choice(boston.shape[0], boston.shape[0], replace=True)
    bootstrap_med.append(boot_med_fn(boston, idx))
print(np.mean(bootstrap_med))
print(np.std(bootstrap_med))
```

The bootstrap_med is very close to the population median.

# (h)

```python
def boot_percentile_fn(data, idx, perc):
    return np.percentile(data.iloc[idx]['medv'], perc)
```

```python
rng = np.random.default_rng(1)
B = 1000 # no. of bootstrap samples
bootstrap_perc_10 = []
for i in range(B):
    idx = rng.choice(boston.shape[0], boston.shape[0], replace=True)
    bootstrap_perc_10.append(boot_percentile_fn(boston, idx, 10))
print(np.mean(bootstrap_perc_10))
print(np.std(bootstrap_perc_10))
```
