# College Dataset

```python
from ISLP import load_data
import pandas as pd
import numpy as np
college = load_data("College")
college
```

```python
college.isna().sum()
```

```python
college.dtypes
```

```python
# Converting the feature Private from Yes/No to 0/1
college["Private"].unique()
```

```python
college["Private"] = college["Private"].map({'Yes':1, 'No':0})
```

# (a)

```python
import sklearn.model_selection as skm
train, test = skm.train_test_split(college, test_size=0.25, shuffle=True, random_state=1)
```

```python
test
```

```python
train
```

```python
X_train, Y_train = train.drop(columns = ["Apps"]), train["Apps"]
X_test, Y_test = test.drop(columns = ["Apps"]), test["Apps"]
```

# (b)

```python
import sklearn.linear_model as skl
model = skl.LinearRegression()
model.fit(X_train, Y_train)
Y_predicted = model.predict(X_test)
```

```python
from sklearn import metrics
metrics.mean_squared_error(Y_test, Y_predicted)
```

# (c)

```python
lambdas = np.logspace(-3, 3, 100)
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
scaler = StandardScaler(with_mean=True, with_std=True)
ridge = skl.RidgeCV(alphas=lambdas)
pipe = Pipeline(steps=[('scaler', scaler), ('ridge', ridge)])
pipe.fit(X_train, Y_train)
```

```python
pipe['ridge'].alpha_
```

```python
Y_predicted_ridge = pipe.predict(X_test)
metrics.mean_squared_error(Y_test, Y_predicted_ridge)
```

# (d)

```python
lasso = skl.LassoCV(alphas=lambdas)
pipe = Pipeline(steps=[('scaler', scaler), ('lasso', lasso)])
pipe.fit(X_train, Y_train)
print(pipe['lasso'].alpha_)
Y_predicted_lasso = pipe.predict(X_test)
print(metrics.mean_squared_error(Y_test, Y_predicted_lasso))
```

```python
pd.DataFrame(pipe['lasso'].coef_, index=X_test.columns)
```

```python
np.count_nonzero(pipe['lasso'].coef_)
```

# (e)

```python
from sklearn.decomposition import PCA
pcr_pipe = Pipeline(steps=[
    ('scaler', StandardScaler()),
    ('pca', PCA()),
    ('linear', skl.LinearRegression())
])
param_grid = {'pca__n_components': range(1, len(X_train.columns) + 1)}
grid = skm.GridSearchCV(
    pcr_pipe,
    param_grid,
    scoring='neg_mean_squared_error',
    cv=5
)
grid.fit(X_train, Y_train)
```

```python
grid.best_params_
```

```python
Y_pred_pcr = grid.predict(X_test)
metrics.mean_squared_error(Y_test, Y_pred_pcr)
```

# (f)

```python
from sklearn.cross_decomposition import PLSRegression
param_grid = {'pls__n_components': range(1, len(X_train.columns) + 1)}
pls_pipe = Pipeline(steps=[('scaler', scaler), ('pls', PLSRegression())])
grid_pls = skm.GridSearchCV(pls_pipe, param_grid, cv = 5)
grid_pls.fit(X_train, Y_train)
```

```python
grid_pls.best_params_
```

```python
Y_pred_pls = grid.predict(X_test)
metrics.mean_squared_error(Y_test, Y_pred_pls)
```

# (g)

Based on the mean squared error observation it is clear, that only Lasso has different error (higher) than other methods.
Since the PCR and PLS didn't reduce the dimension the it looks like all the dimensions are somewhat relevent to the data.
Lasso higher test error suggests that it removed (shrinked) the variable that contained predictive information.
So, we can suggest that the Linear Regression itself is the obvious choice, since it is simple and also explains the data better.
