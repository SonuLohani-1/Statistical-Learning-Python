# (a)

```python
import numpy as np
rng = np.random.default_rng(0)
```

```python
n = 100
X = rng.normal(size=n)
eps = rng.normal(loc = 0.25, scale = 0.5, size=n)
print(X)
print(eps)
```

# (b)

```python
beta0 = 0.325
beta1 = -3
beta2 = 1.5
beta3 = 0.75

Y = beta0 + beta1*X + beta2*np.square(X)+beta3*np.power(X, 3) + eps
Y
```

# (c)

```python
def nCp(sigma2, estimator, X, Y):
    "Negative Cp statistic"
    n, p = X.shape
    Yhat = estimator.predict(X)
    RSS = np.sum((Y - Yhat)**2)
    return -(RSS + 2 * p *sigma2) / n
```

```python
from sklearn.preprocessing import PolynomialFeatures
poly = PolynomialFeatures(degree=10, include_bias=False)
X_poly_array = poly.fit_transform(X.reshape(-1, 1))
X_poly_array
```

```python
import pandas as pd
column_names = [f'X^{i}' for i in range(1, 11)]
X_poly_df = pd.DataFrame(X_poly_array, columns=column_names)
```

```python
from ISLP.models import ModelSpec as MS
design = MS(X_poly_df.columns)
X_design = design.fit_transform(X_poly_df)
X_design
```

```python
from statsmodels.api import OLS
sigma2 = OLS(Y, X_design).fit().scale
sigma2
```

```python
from functools import partial
neg_Cp = partial(nCp, sigma2)
```

```python
from ISLP.models import Stepwise, sklearn_selected
Strategy = Stepwise.fixed_steps(design, n_steps = 10, direction='forward')
```

```python
from statsmodels.api import OLS
selector = sklearn_selected(OLS, Strategy, scoring=neg_Cp)
selector.fit(X_poly_df, Y)
```

```python
selector.selected_state_
```

```python
for state, result in selector.selector_.results_.items():
    print(f"Model Size: {len(state)} -> Negative Cp: {result:.4f}")
```

```python
for state, result in selector.selector_.results_.items():
    if len(state) == 3:
        feature_names = [f.name for f in state]
        print(f"Model Size 3 Features: {feature_names} -> Negative Cp: {result:.4f}")
```
