# (d)

```python
def Power3(x, a):
    return x**a
```

# (e)

```python
x = np.arange(1, 10, 1)
y = Power3(x, 2)
```

```python
fig, ax = plt.subplots()
ax.plot(x, y)
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_xscale('linear')
ax.set_yscale('log')
```

# (f)

```python
def PlotPower(x, a):
    fig, ax = plt.subplots()
    ax.plot(x, Power3(x, a))
    ax.set_xlabel('X')
    ax.set_ylabel('Y')
    ax.set_xscale('linear')
    ax.set_yscale('log')
```

```python
PlotPower(np.arange (1, 11), 3)
```
